```python
import re
from datetime import datetime

def validate_travel_booking(name, passport, phone, email, departure_date, payment_method):


    if not re.fullmatch(r"[A-Za-z\s]+", name):
        return False, print(f"Invalid name: {name}. Only alphabetic characters and spaces are allowed.")
    
    
    if not re.fullmatch(r"[A-Za-z0-9]{9}", passport):
        print(f"Invalid passport number: {passport}. It must be exactly 9 alphanumeric characters.")
    
 
    if not re.fullmatch(r"\d{10}", str(phone)):  
        print(f"Invalid phone number: {phone}. It must be exactly 10 digits.")
    
  
    if not re.fullmatch(r"[^@]+@[^@]+\.[^@]+", email):
        print(f"Invalid email format: {email}.")
    
   
    try:
        datetime.strptime(departure_date, "%d/%m/%Y")
    except ValueError:
        print(f"Invalid departure date: {departure_date}. Format must be 'dd/mm/yyyy'.")
    
 
    if payment_method not in ['credit', 'debit', 'netbanking']:
        print(f"Invalid payment method: {payment_method}. It must be 'credit', 'debit', or 'netbanking'.")

    else:
        return True, print(f"Name: {name} \n Passport: {passport} \n Phone: {phone} \n Email: {email} \n Departure Date: {departure_date} \n Payment Method: {payment_method}")


validate_travel_booking('Anson Thomas', '12345678D', '1234567809', 'anson@gmail.com', '06/09/2024', 'credit')


```
