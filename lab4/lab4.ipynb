```python
from abc import ABC, abstractmethod


class Farm(ABC):
    def __init__(self, name, location):
        self.name = name
        self.location = location

    @abstractmethod
    def display(self):
        pass

class CattleFarm(Farm):
    def __init__(self, name, location, cattle_count, cattlename):
        super().__init__(name, location)
        self.cattle_count = cattle_count
        self.cattlename = cattlename

    def display(self):
        print(f"Farm Name: {self.name}")
        print(f"Location: {self.location}")
        print(f"Cattle Count: {self.cattle_count}")
        print(f"Cattle Names: {self.cattlename}")

class PlantFarm(Farm):
    def __init__(self, name, location, plant_count, plantname):
        super().__init__(name, location)
        self.plant_count = plant_count
        self.plantname = plantname

    def display(self):
        print(f"Farm Name: {self.name}")
        print(f"Location: {self.location}")
        print(f"Plant Count: {self.plant_count}")
        print(f"Plant Names: {self.plantname}")
    
class TreeFarm(Farm):
    def __init__(self, name, location, tree_count, treename):
        super().__init__(name, location)
        self.tree_count = tree_count
        self.treename = treename
        
    def display(self):
        print(f"Farm Name: {self.name}")
        print(f"Location: {self.location}")
        print(f"Tree Count: {self.tree_count}")
        print(f"Tree Names: {self.treename}")


farms = [
    CattleFarm("Sunny Pastures", "Kerala", 80, "Cow, Buffalo, Ox"),
    PlantFarm("Green Fields", "Karnataka", 1000, "Tomatoes"),
    TreeFarm("Mango Grove ", "Andhra", 300, "Mango Tree")
]


for farm in farms:
    farm.display()
    print('\n')  
```

```python
from abc import ABC, abstractmethod

class DataAnalyser(ABC):
    def __init__(self, countAnimal, countPlants, plant, animal):
        self.countPlants = countPlants
        self.countAnimal = countAnimal
        self.plant = plant
        self.animal = animal
        
    @abstractmethod
    def analyse(self):
        pass


class AnalysisError(Exception):
    pass


class TextDataAnalyzer(DataAnalyser):
    def __init__(self, plant, animal):
        super().__init__(0, 0, plant, animal)

    def analyse(self):
        try:
            plant_words = self.plant.split(' ')
            animal_words = self.animal.split(' ')

            self.plant_word_count = len(plant_words)
            self.plant_char_count = len(self.plant)

            self.animal_word_count = len(animal_words)
            self.animal_char_count = len(self.animal)

            print(f'Number of words in plant description: {self.plant_word_count}')
            print(f'Number of words in animal description: {self.animal_word_count}')
            print(f'Number of characters in plant description: {self.plant_char_count}')
            print(f'Number of characters in animal description: {self.animal_char_count}')

        except AttributeError as e:
            print(f"AnalysisError: Invalid data type for text analysis - {e}")
        except ValueError as e:
            print(f"An unexpected value error occurred: {e}")
        except KeyError as e:
            print(f"An unexpected key error occurred: {e}")


class NumericDataAnalyser(DataAnalyser):
    def __init__(self, countAnimal, countPlants):
        super().__init__(countAnimal, countPlants, "", "")

    def analyse(self):
        try:
            total = self.countPlants + self.countAnimal
            print(f'Total number of plants and animals: {total}')
            print(f'Count of animals: {self.countAnimal}')
            print(f'Count of plants: {self.countPlants}')
        except AttributeError as e:
            print(f"AnalysisError: Invalid data type for numeric analysis - {e}")
        except ValueError as e:
            print(f"An unexpected value error occurred: {e}")
        except KeyError as e:
            print(f"An unexpected key error occurred: {e}")




A = [TextDataAnalyzer('Rose Lily Tulip', 'Cow Dog Cat'), NumericDataAnalyser(10, 5)]

for i in A:
    try:
        i.analyse() 
        print('\n') 
    except AnalysisError as e:
        print(f"AnalysisError: {e}")

    
```
